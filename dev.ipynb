## parse pdfs

```python
from pydantic import BaseModel, Field

class InvoiceMetadata(BaseModel):
    invoice_number: str | None = Field(None, description="Invoice number as it appears on the document")
    invoice_date: str | None = Field(None, description="Invoice date in YYYY-MM-DD format")
    irn_number: str | None = Field(None, description="IRN (Invoice Reference Number)")
    acknowledgement_number: str | None = Field(None, description="Acknowledgement number")
    acknowledgement_date: str | None = Field(None, description="Acknowledgement date in YYYY-MM-DD format")
    sale_order_number: str | None = Field(None, description="Sale order number")
    sale_order_date: str | None = Field(None, description="Sale order date in YYYY-MM-DD format")

class Seller(BaseModel):
    name: str | None = Field(None, description="Seller's legal name")
    trade_name: str | None = Field(None, description="Seller's trade name")
    address: str | None = Field(None, description="Seller's address")
    city: str | None = Field(None, description="Seller's city")
    state: str | None = Field(None, description="Seller's state")
    state_code: str | None = Field(None, description="Seller's state code")
    pincode: str | None = Field(None, description="Seller's pincode")
    country: str | None = Field(None, description="Seller's country")
    gstin: str | None = Field(None, description="Seller's GSTIN")
    pan: str | None = Field(None, description="Seller's PAN")
    cin: str | None = Field(None, description="Seller's CIN")
    email: str | None = Field(None, description="Seller's email")
    phone: str | None = Field(None, description="Seller's phone")
    fax: str | None = Field(None, description="Seller's fax")

class Buyer(BaseModel):
    name: str | None = Field(None, description="Buyer's name")
    address: str | None = Field(None, description="Buyer's address")
    city: str | None = Field(None, description="Buyer's city")
    state: str | None = Field(None, description="Buyer's state")
    state_code: str | None = Field(None, description="Buyer's state code")
    pincode: str | None = Field(None, description="Buyer's pincode")
    country: str | None = Field(None, description="Buyer's country")
    gstin: str | None = Field(None, description="Buyer's GSTIN")
    pan: str | None = Field(None, description="Buyer's PAN")
    email: str | None = Field(None, description="Buyer's email")
    phone: str | None = Field(None, description="Buyer's phone")

class Consignee(BaseModel):
    name: str | None = Field(None, description="Consignee's name")
    address: str | None = Field(None, description="Consignee's address")
    city: str | None = Field(None, description="Consignee's city")
    state: str | None = Field(None, description="Consignee's state")
    pincode: str | None = Field(None, description="Consignee's pincode")
    country: str | None = Field(None, description="Consignee's country")
    gstin: str | None = Field(None, description="Consignee's GSTIN")

class GSTDetails(BaseModel):
    gst_customer_type: str | None = Field(None, description="GST customer type: Registered|Unregistered|Composition|Overseas")
    place_of_supply: str | None = Field(None, description="Place of supply")
    reverse_charge_applicable: str | None = Field(None, description="Reverse charge applicable: Yes|No")

class LineItem(BaseModel):
    s_no: int | str | None = Field(None, description="Sequential number or internal item code as shown")
    description: str | None = Field(None, description="Full product/service description")
    parts_code: str | None = Field(None, description="Part code or product code")
    hsn: str | None = Field(None, description="HSN/SAC code")
    quantity: float | None = Field(None, description="Quantity as a number")
    unit: str | None = Field(None, description="Unit of measure (e.g., 'PCS', 'NOS', 'QTY', 'UNITS')")
    unit_price: float | None = Field(None, description="Price per unit before discount and tax")
    discount_percent: float | None = Field(None, description="Percentage discount (0-100)")
    discount_amount: float | None = Field(None, description="Discount amount in currency")
    base_amount: float | None = Field(None, description="Calculated or shown subtotal (qty × unit_price after discount)")
    cgst_rate: float | None = Field(None, description="CGST rate as percentage (e.g., 9, 18)")
    sgst_rate: float | None = Field(None, description="SGST rate as percentage (e.g., 9, 18)")
    igst_rate: float | None = Field(None, description="IGST rate as percentage (e.g., 18, 28)")
    cgst: float | None = Field(None, description="CGST amount in currency")
    sgst: float | None = Field(None, description="SGST amount in currency")
    igst: float | None = Field(None, description="IGST amount in currency")
    total_gst: float | None = Field(None, description="Sum of all GST components")
    amount_incl_gst: float | None = Field(None, description="Total for this line (base + tax)")

class Totals(BaseModel):
    total_line_items_count: int | None = Field(None, description="Total number of line items")
    total_parts_amount: float | None = Field(None, description="Sum of base_amounts before tax")
    total_discount_amount: float | None = Field(None, description="Total discount amount")
    total_taxable_value: float | None = Field(None, description="Total taxable value (total_parts_amount - discounts, or explicit taxable value)")
    cgst_total: float | None = Field(None, description="Total CGST amount")
    sgst_total: float | None = Field(None, description="Total SGST amount")
    igst_total: float | None = Field(None, description="Total IGST amount")
    total_gst: float | None = Field(None, description="Total GST amount")
    other_charges: float | None = Field(None, description="Other charges (freight, packing, etc.)")
    invoice_amount: float | None = Field(None, description="Total invoice amount with tax and other charges")
    invoice_amount_words: str | None = Field(None, description="Invoice amount in words")
    rounding_off_amount: float | None = Field(None, description="Rounding off amount")

class Payment(BaseModel):
    date: str | None = Field(None, description="Payment date in YYYY-MM-DD format")
    amount: float | None = Field(None, description="Payment amount")
    transaction_id: str | None = Field(None, description="Transaction ID")
    payer: str | None = Field(None, description="Payer name")
    mode: str | None = Field(None, description="Payment mode (e.g., 'NEFT', 'CHEQUE', 'Cash')")

class Shipping(BaseModel):
    mode_of_transport: str | None = Field(None, description="Mode of transport")
    vehicle_no: str | None = Field(None, description="Vehicle number")
    lr_gr_no: str | None = Field(None, description="LR/GR number")
    lr_gr_date: str | None = Field(None, description="LR/GR date in YYYY-MM-DD format")
    carrier_name: str | None = Field(None, description="Carrier name")

class DocumentInfo(BaseModel):
    document_type: str | None = Field(None, description="Document type: Tax Invoice|Proforma Invoice|Credit Note|Delivery Challan")
    delivery_note_no: str | None = Field(None, description="Delivery note number")
    delivery_note_date: str | None = Field(None, description="Delivery note date in YYYY-MM-DD format")
    dispatch_date: str | None = Field(None, description="Dispatch date in YYYY-MM-DD format")
    due_date: str | None = Field(None, description="Due date in YYYY-MM-DD format")
    payment_terms: str | None = Field(None, description="Payment terms")

class ExtractionMetadata(BaseModel):
    extraction_timestamp: str = Field(..., description="ISO 8601 timestamp of extraction")
    source_file_name: str = Field(..., description="Source file name")
    extraction_confidence: str | None = Field(None, description="Extraction confidence: 'high', 'medium', 'low'")
    notes: str | None = Field(None, description="Any extraction notes, warnings")

class InvoiceModel(BaseModel):
    invoice_metadata: InvoiceMetadata = Field(..., description="Invoice metadata information")
    # seller: Seller = Field(..., description="Seller information")
    # buyer: Buyer = Field(..., description="Buyer information")
    # consignee: Consignee = Field(..., description="Consignee information")
    gst_details: GSTDetails = Field(..., description="GST details")
    line_items: list[LineItem] = Field(..., description="List of line items")
    totals: Totals = Field(..., description="Invoice totals")
    payments: list[Payment] = Field(..., description="List of payments")
    # shipping: Shipping = Field(..., description="Shipping information")
    document_info: DocumentInfo = Field(..., description="Document information")
    # extraction_metadata: ExtractionMetadata = Field(..., description="Extraction metadata")
```
